# Random Greedy Heuristic Multiple Runs (k = 5)

```python
runs = 20
k = 5
```

## 1 Car

```python
dict = {}
for seed in range(runs):
  network = Network(machine_positions, transport_demand)
  car1 = Car(car_id=1, position=1)
  cars = [car1]
  simulation = Simulation(network, cars, k=k, seed=seed, save=False)
  simulation.start_sim(display=False)
  schedule = simulation.combine_schedules_preserve_order(save=False)
  dict[seed] = (simulation.total_empty_runs, schedule)

best_seed = min(dict, key=lambda x: dict[x][0])
best_empt, best_schedule = dict[best_seed]
best_schedule["emptyRuns"] = best_empt
save_schedule(best_schedule, f"randomized_greedy_heuristic_multistart/schedule_1_MRGH_k{k}_seed{best_seed}.txt")
```

## 5 Cars


```python
dict = {}
for seed in range(runs):
  network = Network(machine_positions, transport_demand)
  cars = [Car(car_id=i, position=i) for i in range(1, 6)]
  simulation = Simulation(network, cars, k=k, seed=seed, save=False)
  simulation.start_sim(display=False)
  schedule = simulation.combine_schedules_preserve_order(save=False)
  dict[seed] = (simulation.total_empty_runs, schedule)

best_seed = min(dict, key=lambda x: dict[x][0])
best_empt, best_schedule = dict[best_seed]
best_schedule["emptyRuns"] = best_empt
save_schedule(best_schedule, f"randomized_greedy_heuristic_multistart/schedule_5_MRGH_k{k}_seed{best_seed}.txt")
```

## 10 Cars

```python
dict = {}
for seed in range(runs):
  network = Network(machine_positions, transport_demand)
  cars = [Car(car_id=i, position=i) for i in range(1, 11)]
  simulation = Simulation(network, cars, k=k, seed=seed, save=False)
  simulation.start_sim(display=False)
  schedule = simulation.combine_schedules_preserve_order(save=False)
  dict[seed] = (simulation.total_empty_runs, schedule)

best_seed = min(dict, key=lambda x: dict[x][0])
best_empt, best_schedule = dict[best_seed]
best_schedule["emptyRuns"] = best_empt
save_schedule(best_schedule, f"randomized_greedy_heuristic_multistart/schedule_10_MRGH_k{k}_seed{best_seed}.txt")
```
